## Data cleaning

```python
# Checking duplicates
df.duplicated()
```

```python
#checking outliers using zscore
from scipy import stats
count =0
z_score = np.abs(stats.zscore(df_new["study_hours_per_day"]))
for z in z_score:
    if z > 3:
        count +=1
print(count)
```

```python
count =0
for i in df_new:
    z_score = np.abs(stats.zscore(df_new[i]))
    for z in z_score:
        if z > 3:
            count +=1
    print(i, " = ", count)
```

```python
X = df.drop(columns = ['productivity_score'])
y = df['productivity_score']

X
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.1, random_state = 42)
```

```python
X_train
```

```python
y_train
```

```python
model = LinearRegression()
model.fit(X_train, y_train)
model.predict(X_test)
```

```python
r2_score = model.score(X_test, y_test)
r2_score
```

### R2_score = 0.9999999673993064

## Saved training data to file

```python
import joblib
joblib.dump(model, "student_productivity_prediction.pkl")
```
