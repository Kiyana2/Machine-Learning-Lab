This project uses my own Student Productivity Prediction API, which was trained and deployed separately using FastAPI and Render. The API receives student information and returns a predicted productivity score.

```python
import requests

data = {
    "gender": 1,
    "study_hours_per_day": 5,
    "sleep_hours": 8,
    "phone_usage_hours": 10,
    "attendance_percentage": 100,
    "stress_level": 8,
    "focus_score": 6
}

response = requests.get(
    "https://student-productivity-prediction-api.onrender.com/predict",
    params=data
)

print(response.json())
```
